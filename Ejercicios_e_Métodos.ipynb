```python
# Importar bibliotecas necesarias
import sympy as sp  # Para manipulación simbólica
import numpy as np  # Para cálculos numéricos
import matplotlib.pyplot as plt  # Para graficar

# Definir el punto de expansión de la serie de Taylor
x0 = 0  # Centro de la expansión
n = 7  # Serie de Taylor de orden 6

# Definir la variable simbólica
x = sp.Symbol('x')

# Obtener la serie de Taylor de cos(x) centrada en x0 con n términos
t = sp.cos(x).series(x, x0, n).removeO()
print(t)  # Imprimir la serie de Taylor

# Convertir la expresión simbólica en una función numérica usable con numpy
t = sp.lambdify(x, t, 'numpy')

# Generar valores de x para la gráfica
xx = np.linspace(x0 - 3, x0 + 3, 1000)  # Rango de valores de x

# Calcular los valores reales de cos(x) y su aproximación con la serie de Taylor
yc = np.cos(xx)  # Valores exactos de cos(x)
yt = t(xx)  # Valores aproximados con la serie de Taylor

# Graficar la función original y su aproximación
plt.plot(xx, yc, label="cos(x)")  # Gráfica de la función original
plt.plot(xx, yt, label="Serie de Taylor orden 7")  # Gráfica de la aproximación

# Agregar leyenda y título
plt.legend()
plt.title("Serie de Taylor de $f(x) = cos(x)$")

# Configuración adicional
plt.grid(False)  # Desactivar la cuadrícula

# Guardar la figura (opcional, descomentarlo si se necesita guardar)
# plt.savefig('seriet.pdf', bbox_inches='tight')

# Mostrar la gráfica
plt.show()
```

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------------------------------------
# 1. Solución Simbólica (Cálculo exacto de la serie)
# -------------------------------------------------------------
# Definimos la variable simbólica x
x = sp.Symbol('x')

# Definimos la función f(x) = e^x
f = sp.exp(x)

# Punto de expansión
x0 = 0
orden = 4

# Obtenemos la serie de Taylor truncada (usando el método .series de SymPy)
# n=5 incluye términos hasta x^4
taylor_series = f.series(x, x0, n=orden + 1).removeO()

print("=== SOLUCIÓN ÁLGEBRAICA ===")
print(f"Función: f(x) = e^x")
print(f"Serie truncada de Taylor de orden {orden} alrededor de x0 = {x0}:")
sp.pprint(taylor_series)


# -------------------------------------------------------------
# 2. Evaluación numérica y gráfica
# -------------------------------------------------------------
# Definición de la función polinómica resultado:
# P4(x) = 1 + x + (x^2)/2 + (x^3)/6 + (x^4)/24
def taylor_e_x_cuarto_orden(x_val):
    return 1 + x_val + (x_val**2)/2 + (x_val**3)/6 + (x_val**4)/24

# Generamos puntos para graficar la función real vs el polinomio de Taylor
x_vals = np.linspace(-2, 2, 400)
y_real = np.exp(x_vals)
y_taylor = taylor_e_x_cuarto_orden(x_vals)

# Graficar
plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_real, label=r'Real: $f(x) = e^x$', color='blue', linewidth=2)
plt.plot(x_vals, y_taylor, label=r'Taylor orden 4: $P_4(x)$', color='red', linestyle='--')
plt.title(r'Aproximación de Taylor de orden 4 para $e^x$ en $x_0 = 0$')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', linewidth=0.8, linestyle=':')
plt.axvline(0, color='black', linewidth=0.8, linestyle=':')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

**PROGRAMA INGENIERIA AGROINDUSTRIAL**


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------------------------------------
# 1. Solución Simbólica (Cálculo exacto de la serie)
# -------------------------------------------------------------
# Definimos la variable simbólica x
x = sp.Symbol('x')

# Definimos la función f(x) = e^x
f = sp.exp(x)

# Punto de expansión
x0 = 0
orden = 4

# Obtenemos la serie de Taylor truncada (usando el método .series de SymPy)
# n=5 incluye términos hasta x^4
taylor_series = f.series(x, x0, n=orden + 1).removeO()

print("=== SOLUCIÓN ÁLGEBRAICA ===")
print(f"Función: f(x) = e^x")
print(f"Serie truncada de Taylor de orden {orden} alrededor de x0 = {x0}:")
sp.pprint(taylor_series)


# -------------------------------------------------------------
# 2. Evaluación numérica y gráfica
# -------------------------------------------------------------
# Definición de la función polinómica resultado:
# P4(x) = 1 + x + (x^2)/2 + (x^3)/6 + (x^4)/24
def taylor_e_x_cuarto_orden(x_val):
    return 1 + x_val + (x_val**2)/2 + (x_val**3)/6 + (x_val**4)/24

# Generamos puntos para graficar la función real vs el polinomio de Taylor
x_vals = np.linspace(-2, 2, 400)
y_real = np.exp(x_vals)
y_taylor = taylor_e_x_cuarto_orden(x_vals)

# Graficar
plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_real, label=r'Real: $f(x) = e^x$', color='blue', linewidth=2)
plt.plot(x_vals, y_taylor, label=r'Taylor orden 4: $P_4(x)$', color='red', linestyle='--')
plt.title(r'Aproximación de Taylor de orden 4 para $e^x$ en $x_0 = 0$')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', linewidth=0.8, linestyle=':')
plt.axvline(0, color='black', linewidth=0.8, linestyle=':')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------------------------------------
# 1. Solución Simbólica con SymPy
# -------------------------------------------------------------
x = sp.Symbol('x')

# Definimos la función f(x) = ln(x)
f = sp.log(x)

x0 = 1
orden = 5

# Obtenemos la serie de Taylor truncada
taylor_series = f.series(x, x0, n=orden + 1).removeO()

print("=== RESULTADO SIMBÓLICO ===")
print(f"Serie truncada de Taylor de orden {orden} para ln(x) en x0 = {x0}:")
sp.pprint(taylor_series)

# -------------------------------------------------------------
# 2. Evaluación numérica y representación gráfica
# -------------------------------------------------------------
# Función polinómica resultado
def taylor_ln_5(x_val):
    u = x_val - 1
    return u - (u**2)/2 + (u**3)/3 - (u**4)/4 + (u**5)/5

# Rango de valores cercanos a x0 = 1 para graficar (evitando x <= 0)
x_vals = np.linspace(0.2, 2.0, 400)
y_real = np.log(x_vals)
y_taylor = taylor_ln_5(x_vals)

# Graficar
plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_real, label=r'Real: $f(x) = \ln(x)$', color='blue', linewidth=2)
plt.plot(x_vals, y_taylor, label=r'Taylor orden 5: $P_5(x)$', color='red', linestyle='--')
plt.plot(x0, np.log(x0), 'ro', label=f'Punto $x_0 = {x0}$')

plt.title(r'Aproximación de Taylor de 5to orden para $\ln(x)$ en $x_0 = 1$')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', linewidth=0.8, linestyle=':')
plt.axvline(1, color='gray', linewidth=0.8, linestyle=':')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

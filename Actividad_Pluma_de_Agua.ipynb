```python
# # 1. Solicitar al usuario el número de plumas
numero_plumas = float(input("Ingrese la cantidad de plumas de agua recibidas: "))

# # 2. Definir constantes del problema
# Usamos los datos exactos de la imagen suministrada
CAUDAL_POR_PLUMA_L_S = 0.025  # 1 pluma = 0.025 L/s
SEGUNDOS_EN_UN_DIA = 86400    # 24 horas * 3600 segundos
DIAS_DEL_MES = 30

# # 3. Calcular el caudal total
caudal_total_l_s = numero_plumas * CAUDAL_POR_PLUMA_L_S

# # 4. Calcular el consumo diario
# Pasamos de litros por segundo (L/s) a litros por día multiplicando por los segundos del día
consumo_diario_litros = caudal_total_l_s * SEGUNDOS_EN_UN_DIA

# # 5. Calcular el consumo mensual
consumo_mensual_litros = consumo_diario_litros * DIAS_DEL_MES

# # 6. Clasificar el consumo
# Clasificación según los rangos de la imagen basados en el consumo diario
if consumo_diario_litros < 1000:
    clasificacion = "Bajo"
elif 1000 <= consumo_diario_litros <= 3000:
    clasificacion = "Medio"
else:
    clasificacion = "Alto"

# # 7. Mostrar resultados
print("\n" + "="*45)
print("        REPORTE OFICIAL DE CONSUMO")
print("="*45)
print(f"Cantidad de plumas ingresadas : {numero_plumas}")
print(f"1. Caudal total               : {caudal_total_l_s:.3f} L/s")
print(f"2. Consumo diario             : {consumo_diario_litros:.2f} litros")
print(f"3. Consumo mensual (30 días)  : {consumo_mensual_litros:.2f} litros")
print(f"4. Clasificación del consumo  : {clasificacion}")
print("="*45)
```
